```python
import numpy as np
import matplotlib.pyplot as plt

# Problema 1

c_min = -2
c_max = 1
cantidad_c = 1000
c_vals = np.linspace(c_min, c_max, cantidad_c)

x_min = -2
x_max = 2
cantidad_x = 1000

histograma = np.zeros((cantidad_c, cantidad_x))

for i in range(cantidad_c):
    c = c_vals[i]
    x = 0

    for n in range(200):
        x = x**2 + c

        if n >= 100:
            if x_min <= x <= x_max:
                j = int((x - x_min) / (x_max - x_min) * cantidad_x)
                j = min(j, cantidad_x - 1)
                histograma[i, j] += 1


plt.imshow(histograma.T, extent=[c_min, c_max, x_min, x_max], cmap='inferno', origin='lower', aspect='auto')
plt.title("Diagrama de Órbitas para $Q_c(x) = x^2 + c$")
plt.xlabel("Valor de c")
plt.ylabel("Valores de x (atractores)")
plt.colorbar(label="Frecuencia")
plt.show()
```

```python
# Problema 2

ancho = 300
alto = 300
max_iter = 100

imagen = []

for y in range(alto):
    fila = []
    for x in range(ancho):
        re = -2 + (x / ancho) * 3
        im = -1.5 + (y / alto) * 3
        c = complex(re, im)

        z = 0
        iteraciones = 0

        while abs(z) <= 2 and iteraciones < max_iter:
            z = z**2 + c
            iteraciones += 1

        gris = int(255 * iteraciones / max_iter)
        fila.append(255 - gris)

    imagen.append(fila)

plt.imshow(imagen, cmap='gray', extent=[-2, 1, -1.5, 1.5])
plt.title("Conjunto de Mandelbrot")
plt.xlabel("Re(c)")
plt.ylabel("Im(c)")
plt.show()
```
